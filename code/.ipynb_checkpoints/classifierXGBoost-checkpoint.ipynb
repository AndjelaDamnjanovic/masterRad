```python
import os
from Bio import SeqIO
from sklearn.preprocessing import LabelEncoder
from collections import Counter
import xgboost as xgb
from sklearn.metrics import classification_report, accuracy_score

# --- Load repeats info ---
files = []
folder = "../proba2NoDuplicates/direct"
for file in os.listdir(folder):
    if file.startswith("."):
        continue
    files.append(folder + "/" + file)

labels = []
repeats = []
for file in files:
    with open(file, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            if len(parts) >= 2:
                labels.append(parts[0])
                repeats.append(parts[1])

known_repeats = list(set(repeats))
print(f"Number of unique repeats: {len(known_repeats)}")

def extract_repeat_counts(sequence, repeat_list):
    return [sequence.count(r) for r in repeat_list]

# --- Stream training sequences and extract features on the fly ---
X_train = []
y_train = []
train_folder = "../trainSetNoDuplicates"

for file in os.listdir(train_folder):
    if file.startswith("."):
        continue
    filepath = os.path.join(train_folder, file)
    for record in SeqIO.parse(filepath, "fasta"):
        X_train.append(extract_repeat_counts(str(record.seq), known_repeats))
        y_train.append(file.replace("../trainSetNoDuplicates/", "").replace(".fasta", ""))

# --- Stream testing sequences and extract features on the fly ---
X_test = []
y_test = []
test_folder = "../testSetNoDuplicates"

for file in os.listdir(test_folder):
    if file.startswith("."):
        continue
    filepath = os.path.join(test_folder, file)
    for record in SeqIO.parse(filepath, "fasta"):
        X_test.append(extract_repeat_counts(str(record.seq), known_repeats))
        y_test.append(file.replace("../testSetNoDuplicates/", "").replace(".fasta", ""))

# --- Encode labels ---
encoder = LabelEncoder()
y_train_enc = encoder.fit_transform(y_train)
y_test_enc = encoder.transform(y_test)

# --- Calculate sample weights ---
counter = Counter(y_train_enc)
total = sum(counter.values())
weights = {cls: total / count for cls, count in counter.items()}
sample_weights = [weights[label] for label in y_train_enc]

# --- Prepare DMatrix ---
dtrain = xgb.DMatrix(X_train, label=y_train_enc, weight=sample_weights)
dtest = xgb.DMatrix(X_test, label=y_test_enc)

# --- Train model ---
num_classes = len(encoder.classes_)
params = {
    "objective": "multi:softprob",
    "num_class": num_classes,
    "eval_metric": "mlogloss",
    "seed": 42,
}
model = xgb.train(params, dtrain, num_boost_round=100)

# --- Predictions ---
y_pred_probs = model.predict(dtest)
y_pred_enc = y_pred_probs.argmax(axis=1)
y_pred = encoder.inverse_transform(y_pred_enc)

y_pred_probs_train = model.predict(dtrain)
y_pred_enc_train = y_pred_probs_train.argmax(axis=1)
y_pred_train = encoder.inverse_transform(y_pred_enc_train)

# --- Report ---
report = classification_report(y_test, y_pred)
print("Classification report:")
print(report)
print("Accuracy score test: " + str(accuracy_score(y_test, y_pred)))
print("Accuracy score train: " + str(accuracy_score(y_train, y_pred_train)))

with open("classification_report_xgb_direct_no_duplicates.txt", "w") as f:
    f.write(report)
    f.write("\nAccuracy score test: ")
    f.write(str(accuracy_score(y_test, y_pred)))
    f.write("\nAccuracy score train: ")
    f.write(str(accuracy_score(y_train, y_pred_train)))
```
