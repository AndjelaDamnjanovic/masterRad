```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, accuracy_score
from sklearn.feature_extraction.text import CountVectorizer

import matplotlib.pyplot as plt
from collections import Counter

from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO

import os
import pandas as pd

import numpy as np
from imblearn.over_sampling import SMOTE
```

```python
# -----------------------------
# 1️⃣ Load known repeats
files = []
folder = "../proba2NoDuplicates/direct"
for file in os.listdir(folder):
    if file.startswith("."):
        continue
    files.append(folder + "/" + file)

labels = []
repeats = []

for file in files:
    with open(file, "r") as f:
        for line in f:
            parts = line.strip().split(",")
            if len(parts) >= 2:
                labels.append(parts[0])
                repeats.append(parts[1])

known_repeats = list(set(repeats))
print(f"Number of unique repeats: {len(known_repeats)}")

# -----------------------------
# 2️⃣ Load training data
files = []
folder = "../trainSetNoDuplicates"
for file in os.listdir(folder):
    if file.startswith("."):
        continue
    files.append(folder + "/" + file)

labelTrain = []
sequencesTrain = []

for file in files:
    for record in SeqIO.parse(file, "fasta"):
        sequencesTrain.append(str(record.seq))
        labelTrain.append(file.replace("../trainSetNoDuplicates/", "").replace(".fasta", ""))

# -----------------------------
# 3️⃣ Load test data
files = []
folder = "../testSetNoDuplicates"
for file in os.listdir(folder):
    if file.startswith("."):
        continue
    files.append(folder + "/" + file)

labelTest = []
sequencesTest = []

for file in files:
    for record in SeqIO.parse(file, "fasta"):
        sequencesTest.append(str(record.seq))
        labelTest.append(file.replace("../testSetNoDuplicates/", "").replace(".fasta", ""))

# -----------------------------
# 4️⃣ Feature extraction
def extract_repeat_counts(sequence, repeat_list):
    return [sequence.count(r) for r in repeat_list]

X_train = [extract_repeat_counts(seq, known_repeats) for seq in sequencesTrain]
X_test = [extract_repeat_counts(seq, known_repeats) for seq in sequencesTest]

# -----------------------------
# 5️⃣ Encode labels
encoder = LabelEncoder()
y_train_enc = encoder.fit_transform(labelTrain)
y_test_enc = encoder.transform(labelTest)

# -----------------------------
# 6️⃣ Filter out small classes (fewer than 3 samples)
counts = Counter(y_train_enc)
keep_classes = {cls for cls, count in counts.items() if count >= 3}
print(f"Keeping {len(keep_classes)} classes with ≥3 samples.")

train_mask = np.array([y in keep_classes for y in y_train_enc])
X_train = np.array(X_train)[train_mask]
y_train_enc = np.array(y_train_enc)[train_mask]

test_mask = np.array([y in keep_classes for y in y_test_enc])
X_test = np.array(X_test)[test_mask]
y_test_enc = np.array(y_test_enc)[test_mask]

# -----------------------------
# 7️⃣ Apply SMOTE
min_class_size = min(Counter(y_train_enc).values())
k_neighbors = max(1, min(5, min_class_size - 1))
print(f"Applying SMOTE with k_neighbors = {k_neighbors}")

smote = SMOTE(random_state=42, k_neighbors=k_neighbors)
X_train_res, y_train_enc_res = smote.fit_resample(X_train, y_train_enc)

# -----------------------------
# 8️⃣ Train Naive Bayes
nb = MultinomialNB()
nb.fit(X_train_res, y_train_enc_res)

# -----------------------------
# 9️⃣ Predict & Evaluate
y_pred_enc = nb.predict(X_test)
y_pred = encoder.inverse_transform(y_pred_enc)
y_true = encoder.inverse_transform(y_test_enc)

y_pred_enc_train = nb.predict(X_train)
y_pred_train = encoder.inverse_transform(y_pred_enc_train)
y_true_train = encoder.inverse_transform(y_train_enc)

report = classification_report(y_true, y_pred)

print("Classification report: ")
print(report)

print("Accuracy score test: " + str(accuracy_score(y_true, y_pred)))
print("Accuracy score train: " + str(accuracy_score(y_true_train, y_pred_train)))

with open("classification_report_naiveBayes_direct_SMOTE_no_duplicates_no_small_classes.txt", "w") as f:
    f.write(report)
    f.write("\nAccuracy score test: ")
    f.write(str(accuracy_score(y_true, y_pred)))
    f.write("\nAccuracy score train: ")
    f.write(str(accuracy_score(y_true_train, y_pred_train)))
```

```python
import os
from collections import Counter
import numpy as np
from Bio import SeqIO
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score
from imblearn.over_sampling import SMOTE

# -----------------------------
# 1️⃣ Load known repeats
def load_known_repeats(folder):
    known = set()
    for file in os.listdir(folder):
        if not file.startswith("."):
            with open(os.path.join(folder, file), "r") as f:
                for line in f:
                    parts = line.strip().split(",")
                    if len(parts) >= 2:
                        known.add(parts[1])
    return list(known)

# -----------------------------
# 2️⃣ Stream sequences and labels
def stream_sequences(folder):
    for file in os.listdir(folder):
        if not file.startswith("."):
            label = file.replace(".fasta", "")
            for record in SeqIO.parse(os.path.join(folder, file), "fasta"):
                yield str(record.seq), label

# -----------------------------
# 3️⃣ Feature extraction
def extract_repeat_counts(sequence, repeat_list):
    return [sequence.count(r) for r in repeat_list]

# -----------------------------
# 4️⃣ Load known repeats
known_repeats = load_known_repeats("proba2/DC")
print(f"Number of unique repeats: {len(known_repeats)}")

# -----------------------------
# 5️⃣ Stream & process training data
X_train, y_train = [], []
for seq, label in stream_sequences("trainSet"):
    X_train.append(extract_repeat_counts(seq, known_repeats))
    y_train.append(label)

# -----------------------------
# 6️⃣ Stream & process test data
X_test, y_test = [], []
for seq, label in stream_sequences("testSet"):
    X_test.append(extract_repeat_counts(seq, known_repeats))
    y_test.append(label)

# -----------------------------
# 7️⃣ Encode labels
encoder = LabelEncoder()
y_train_enc = encoder.fit_transform(y_train)
y_test_enc = encoder.transform(y_test)

# -----------------------------
# 8️⃣ Remove classes with <3 samples
counts = Counter(y_train_enc)
keep_classes = {cls for cls, count in counts.items() if count >= 3}
print(f"Keeping {len(keep_classes)} classes with ≥3 samples.")

train_mask = np.array([y in keep_classes for y in y_train_enc])
X_train = np.array(X_train)[train_mask]
y_train_enc = np.array(y_train_enc)[train_mask]

test_mask = np.array([y in keep_classes for y in y_test_enc])
X_test = np.array(X_test)[test_mask]
y_test_enc = np.array(y_test_enc)[test_mask]

# -----------------------------
# 9️⃣ SMOTE
min_class_size = min(Counter(y_train_enc).values())
k_neighbors = max(1, min(5, min_class_size - 1))
print(f"Applying SMOTE with k_neighbors = {k_neighbors}")
smote = SMOTE(random_state=42, k_neighbors=k_neighbors)
X_train_res, y_train_enc_res = smote.fit_resample(X_train, y_train_enc)

# -----------------------------
# 🔟 Train & evaluate Naive Bayes
nb = MultinomialNB()
nb.fit(X_train_res, y_train_enc_res)

y_pred_enc = nb.predict(X_test)
y_pred = encoder.inverse_transform(y_pred_enc)
y_true = encoder.inverse_transform(y_test_enc)

print("Classification report:")
print(classification_report(y_true, y_pred))
print("Accuracy score:", accuracy_score(y_true, y_pred))
```
