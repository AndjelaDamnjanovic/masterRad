```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Load your images (replace with your actual paths)
image_paths = ["masterRadDC.png", "masterRadDN.png", "masterRadIC.png", "masterRadIN.png"]
captions = ["a)", "b)", "c)", "d)"]
images = [mpimg.imread(path) for path in image_paths]

# Create 2x2 subplot grid
fig, axs = plt.subplots(2, 2, figsize=(6, 4))
axs = axs.flatten()

# Display each image in a subplot
for ax, img, caption in zip(axs, images, captions):
    ax.imshow(img)
    ax.axis('off')  # Hide axis if desired
    ax.set_title(caption, fontsize=8, y=-0.75)

plt.tight_layout()
plt.show()
```
