```python
from Bio import SeqIO
import os
import csv
```

```python
def createFolder(folder):
    if not os.path.exists(folder):
        os.makedirs(folder)
        print(f"Folder '{folder}' created.")
    else:
        print(f"Folder '{folder}' already exists.")
```

```python
createFolder("../statistics")
```

```python
def getStatistics(inFile, outFile):
    mapping = {
        'equine-CoV' : 1363,
        'turkey-CoV' : 1226,
        'porcine-CoV': 1160,
        'bat-CoV': 1259,
        'IBV': 1205,
        'rat-CoV': 1365,
        'canine-CoV': 1482,
        'feline-CoV': 1472,
        'MERS-CoV' : 1366,
        'hedgehog-CoV' : 1330,
        'HCoV-229E': 1582,
        'HCoV-NL63': 1356,
        'dolphin-CoV': 1495,
        'SARS-CoV': 1255,
        'ferret-CoV': 1457,
        'SARS-CoV2': 1276,
        'bovine-CoV' : 1376,
        'HCoV-HKU1': 1359,
        'rabbit-CoV': 1363,
        'HCoV-OC43': 1366     
    }
    lenSpike = 0
    numInstances = 0

    lenSpikeWhole = 0
    numInstancesWhole = 0

    type = inFile.replace("../aaFastaFiles/", "").replace(".fasta", "")
    key = mapping[type]
    
    for record in SeqIO.parse(inFile, "fasta"):
        lenSeq = len(record.seq)
        lenSpike += lenSeq
        numInstances += 1

        if lenSeq in range(key - 10, key + 10):
            lenSpikeWhole += lenSeq
            numInstancesWhole += 1
    data = [[type, str(lenSpike/numInstances), str(lenSpikeWhole/numInstancesWhole), numInstances]]
    with open(outFile, 'a', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerows(data)
```

```python
def makeCSVfile(filename):
    headers = ["Coronavirus type", "Avg spike len", "Avg complete spike len", "Num instances"]
    with open(filename, mode="a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(headers)
```

```python
baseFolder = "../aaFastaFiles/"
csvFile = "../statistics/statistics.csv"
makeCSVfile(csvFile)

for file in os.listdir(baseFolder):
    if file.startswith("."):
        continue
    path = baseFolder + file
    getStatistics(path, csvFile)
```
